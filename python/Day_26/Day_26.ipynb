```python
from datetime import datetime

# Accept input from user
day = int(input("Enter day: "))
mon = input("Enter month name (e.g., March): ")
year = int(input("Enter year: "))

# Convert month name to month number
month_number = datetime.strptime(mon, "%B").month

# Create datetime object
d_obj = datetime(year, month_number, day)

# Print formatted outputs
print(f"Date object: {d_obj}")
print(f"Day of the week: {d_obj.strftime('%A')}")
print(f"Month name: {d_obj.strftime('%B')}")

# Alternative input format: "day,month,year"
d, m, y = map(int, input("Enter date as day,month,year: ").split(","))
d_obj2 = datetime(y, m, d)

print(f"Date object: {d_obj2}")
print(f"Day of the week: {d_obj2.strftime('%A')}")
print(f"Month name: {d_obj2.strftime('%B')}")
```
