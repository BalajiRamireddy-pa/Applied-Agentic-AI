```python
class Students:
    """Student detail AAA batch"""

    # Class attribute
    batch = "AAA-HYD_001"

    # Method to take student details
    def stu_data(self):
        self.name = input("Enter student name: ")
        self.age = int(input("Enter age: "))
        self.place = input("Enter place: ")

    # Method to display student details
    def details(self):
        print(f"Student name: {self.name}")
        print(f"Student from: {self.place} and age: {self.age}")


# Creating an object of Students class
Std1 = Students()

# Accessing class attribute
print(Std1.batch)

# Taking student details
Std1.stu_data()

# Displaying student details
Std1.details()

# Displaying instance attributes
print(Std1.__dict__)

# Displaying class documentation
print(Std1.__doc__)

# Displaying class information
print(Std1.__class__)

# Taking student details again
Std1.stu_data()

# Displaying updated student details
Std1.details()
```
