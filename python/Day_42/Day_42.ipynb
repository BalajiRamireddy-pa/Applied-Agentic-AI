```python
# data  visulization -->matplotlib

# line plot --> serial repenstaion data

months=["Jan","Feb","Mar","Apr","May"]
sales = [2500,3000,2000,3500,2600]
# now we will use plot()
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [2500, 3000, 2000, 3500, 2600]

plt.plot(months, sales, color='red', marker='o')

plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")

plt.show()
```

```python
# subplot

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales1 = [2500, 3000, 2000, 3500, 2600]
profit = [25,15,35,45,20]
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Sales Analysis")
plt.plot(months, sales1,)
plt.subplot(1,2,2)
plt.plot(months, profit)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales and Profit ")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales1 = [2500, 3000, 2000, 3500, 2600]
profit = [25, 15, 35, 45, 20]

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(months, sales1)
plt.title("Sales")

plt.subplot(2, 2, 2)
plt.plot(months, profit)
plt.title("Profit")

plt.subplot(2, 2, 3)
plt.bar(months, sales1)
plt.title("Sales Bar")

plt.subplot(2, 2, 4)
plt.bar(months, profit)
plt.title("Profit Bar")

plt.show()
```
