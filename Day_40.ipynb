```python
# Numpy arr vs list (Faster computation)

# let us creat list of 2

l1 = list(range(10))
l2 = list(range(10))

print(l1+l2) # merging 


# time module for time calcualtion
import time 
start = time.time()
l3 = [x+y for x,y in zip(l1,l2)]
print(l3)
```

```python
l3 = list(range(1000000))
l4 = list(range(1000000))


start = time.time()
l4 = [x+y for x,y in zip(l3,l4)]
end = time.time()

print(f"Excution Time for List is {(end-start)*1000} milliseconds")
```

```python
# creating ele in numpy arr --> arrange()

A1 = np.arange(1000000)
A2 = np.arange(1000000)
start = time.time()
A3 = A1+A2
end = time.time()
print(f"Excution Time for List is {(end-start)*1000} milliseconds")
```
