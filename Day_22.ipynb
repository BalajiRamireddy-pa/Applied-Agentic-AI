```python
# list comprehension

# append ele in list
lst = []
for i in range(10):
    lst.append(i)
    print(lst)

# append ele in list with comprehension
lst = [i for i in range(10)]
print(lst)

# operations
n = int(input())
lst = [i for i in range(n)]
print(lst)
lst.append(56)
print(lst)
print(lst[::-1])
```

```python
# squares of numbers

data = [i ** 2 for i in range(10)]
print(data)

e = [i % 2 == 1 for i in range(10)]
print(e)

details = ['saketh', 'codegnan', 'data', 'agents', 'rag']
new = [i.upper() for i in details]
print(new)
print(*new)

a, *nam, c = 1, 'saketh', 'codegnan', 'data', 34
print(a)
print(nam)  # output as -> ['saketh', 'codegnan', 'data']
print(*nam)  # output as -. saketh codegnan data
print(c)

a = [15, 20, 25, 30]
# update val in list
a = [i + 5 for i in a]
print(a)
```

```python
# list comprehension using if-else cond
data = [12, 3, 4, 5, 6, 7, 9]
print(data)
res = ["New" if i % 2 == 0 else "old" for i in data]
print(res)

# Nested List Condition

a = [(i, j) for i in range(5) for j in range(3)]
print(a)

b = [(i, j) for i in [1, 2, 3, 4] for j in [2, 3, 5]]
print(b)
```

```python
# multiplication
b = [f"{i}*{j}= {i * j}" for i in range(1, 11) for j in range(1, 11)]
print(b)
print(*b)
```

```python
# colors
colors = ['Red', 'Blue', 'Green']
sizes = ["S", "M", "L"]
dress = [(i, j) for i in colors for j in sizes]
print(dress)

# Nested comprehension with if
b = [i * j for i in range(1, 11) for j in range(1, 11) if i != j]
print(b)
print(*b)

# using if-else in list comprehension
a = [1, 3, 5, 7]
b = [2, 4, 6, 8, 9]
c = [x + 5 if x < y else x for x in a for y in b]
print(c)

a = [1, 3, 5, 7]
b = [2, 4, 6, 8, 9]
c = (x + 5 if x < y else x for x in a for y in b)
print(c)
```

```python

# no tuple comprehension - generator

"""Normal syntax """


def fun():
    return 1, 2, 3, 4


print(fun())


# yield --> generator functions
def fun():
    yield 1
    yield 2
    yield 3


b = fun()
print(next(b))
print(next(b))


def display():
    yield "Python"
    yield "GenAi"
    yield "Rag"
    yield "Agentic"


print(display())
print(type(display()))
d = display()
print(next(d))
```
