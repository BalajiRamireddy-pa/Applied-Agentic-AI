```python
# Library

from google import genai # genai lib acessing LLM
import pathlib # acessing the Api key from sepfic path
import textwrap # used to format text
from IPython.display import display # use to display img
from IPython.display import Markdown # used to display img
```

```python
# func to formate the text

def to_markdown(text):
  text = text.replace('..','- ') # in the text replacing ... with Bullets piont
  return Markdown(text)
```

```python
# Config api key

from google.colab import userdata # to acess the api key from colab
GOOGLE_API = userdata.get("gemin-api") # acessing the api key inside our code

client = genai.Client(api_key=GOOGLE_API) # creating a client obj to acess llms
```

```python
# list all the google llm with client obj

for m in client.models.list():
  print(m.name)
```

```python
# Task-1 : Generate the text using llm

response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = "What is meaning of life"
)
print(response)
```

```python
to_markdown(response.text)
```

```python
from google.genai import types

with open("image.jpg", "rb") as img_file:
    img_data = img_file.read()

img_part = types.Part.from_bytes(
    data=img_data,
    mime_type="image/jpeg"
)
```

```python
response = client.models.generate_content(
        model="gemini-3.6-flash",
    contents=["Describe this image...",img_part]

)
to_markdown(response.text)
```
