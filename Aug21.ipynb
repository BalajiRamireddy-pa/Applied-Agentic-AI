```python
#Numpy Arrays vs Lists (Fast Computataion )
import time 
#Let us create two Lists with Around 10,00,000 elements 
L1=list(range(1000000))
L2=list(range(1000000))
#print(L1+L2)   #Merging 
#we will Use our time module to get time calculation 
start= time.time()
L3=[x+y for x,y in zip(L1,L2)]
#print(L3)
end = time.time()
print(end-start)
print(f'Execution Time for Lists is {(end-start)*1000} milliseconds ')
```

```python
#Create  elements in Numpy array Use --- >Arange()
A1=np.arange(1000000)
A2=np.arange(1000000)
start = time.time()
A3=A1+A2
end= time.time()
print(end-start)
print(f'Execution time  for Numpy Array is {(end-start)*1000} milliseconds ')
```

```python
#Indexing and Slicing on Arrays
data=np.array([[1,4,5],[5,6,-9],[2,3,-1]])
print(data)
print(data[0])
print(data[1])
print(data[2])
print(data[0][1])
print(data[0:,0])
print(data[1:,1:])
print(data[0:2,2])
print(data[1:2,2])
```

```python
#Range of Number ---> arange(),linspace(),  -->return evenly spaced values in Intervals
a=np.linspace(1,10)   #by defaults interval=50
a
a=np.linspace(1,10,4)   #num=4 Marks 4 values to be specified 
a
a=np.linspace(1,10,4, endpoint=True,retstep=True,dtype=int)
a
```
